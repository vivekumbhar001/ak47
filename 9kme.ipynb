```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv("dataset_09.csv", header=None)
```

```python
# Select features (exclude last column: attack label)
X = df.iloc[:, :-1]
```

```python
# Convert categorical data into numeric values
X = pd.get_dummies(X)
X.columns = X.columns.astype(str)
```

```python
# Replace infinite values and fill missing values
X = X.replace([float("inf"), float("-inf")], float("nan"))
X = X.fillna(X.median())
```

```python
# Normalize data
X_scaled = StandardScaler().fit_transform(X)
```

```python
# Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X_scaled)
```

```python
# Calculate distance from cluster centers
dist = ((X_scaled - kmeans.cluster_centers_[df["Cluster"]])**2).sum(axis=1)**0.5

# Detect unusual traffic
threshold = pd.Series(dist).quantile(0.95)
df["Traffic"] = ["Unusual" if d > threshold else "Normal" for d in dist]
```

```python
# Display output
print(df[["Cluster", "Traffic"]].head(10))
print("Unusual traffic:", (df["Traffic"] == "Unusual").sum())
```
